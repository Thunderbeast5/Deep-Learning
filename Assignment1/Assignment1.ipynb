```python
# 1. Import Libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 2. AND Gate Dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])
```

```python
# 3. Initialize Parameters
weights = np.zeros(2)
bias = 0
learning_rate = 0.1
epochs = 100
```

```python
# 4. Step Activation Function
def step(z):
    return 1 if z >= 0 else 0
```

```python
# 5. Train Perceptron
for epoch in range(epochs):

    for i in range(len(X)):

        # Calculate weighted sum
        z = np.dot(X[i], weights) + bias

        # Predict output
        prediction = step(z)

        # Calculate error
        error = y[i] - prediction

        # Update weights and bias
        weights = weights + learning_rate * error * X[i]
        bias = bias + learning_rate * error
```

```python
# 6. Display Predictions
predictions = []

for i in range(len(X)):
    z = np.dot(X[i], weights) + bias
    predictions.append(step(z))

print("Input Data:")
print(X)

print("\nActual Outputs:")
print(y)

print("\nPredicted Outputs:")
print(predictions)

print("\nFinal Weights:", weights)
print("Final Bias:", bias)
```

```python
# 7. Visualize Decision Boundary

plt.figure(figsize=(7, 5))

# Plot data points
plt.scatter(X[y == 0, 0], X[y == 0, 1],
            color='red', label='Class 0')

plt.scatter(X[y == 1, 0], X[y == 1, 1],
            color='blue', marker='x', s=100,
            label='Class 1')

# Calculate decision boundary
x1 = np.linspace(-0.5, 1.5, 100)
x2 = -(weights[0] * x1 + bias) / weights[1]

plt.plot(x1, x2, 'k--', label='Decision Boundary')

# Graph details
plt.xlabel("Input 1 (X1)")
plt.ylabel("Input 2 (X2)")
plt.title("Perceptron Decision Boundary for AND Gate")
plt.legend()
plt.grid(True)
plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)

plt.show()
```
