```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load dataset from CSV
df = pd.read_csv('perceptron.csv')
print(df)

X = df[['X1', 'X2']].to_numpy(dtype=float)
y = df['Output'].to_numpy(dtype=int)
```

```python
# Initialize perceptron parameters
weights = np.zeros(2)
bias = 0
learning_rate = 0.1
epochs = 100
```

```python
# Step activation function
def step(z):
    return 1 if z >= 0 else 0
```

```python
# Train the perceptron
for epoch in range(epochs):
    for i in range(len(X)):
        z = np.dot(X[i], weights) + bias
        prediction = step(z)
        error = y[i] - prediction

        if error != 0:
            weights = weights + learning_rate * error * X[i]
            bias = bias + learning_rate * error
```

```python
# Display predictions
predictions = []
for i in range(len(X)):
    z = np.dot(X[i], weights) + bias
    predictions.append(step(z))

print('Input Data:')
print(X)
print('\nActual Outputs:')
print(y)
print('\nPredicted Outputs:')
print(predictions)
print('\nFinal Weights:', weights)
print('Final Bias:', bias)
```

```python
# Visualize the decision boundary
plt.figure(figsize=(7, 5))
plt.scatter(X[y == 0, 0], X[y == 0, 1], color='red', label='Class 0')
plt.scatter(X[y == 1, 0], X[y == 1, 1], color='blue', marker='x', s=100, label='Class 1')

x1 = np.linspace(-0.5, 1.5, 100)
x2 = -(weights[0] * x1 + bias) / weights[1]
plt.plot(x1, x2, 'k--', label='Decision Boundary')

plt.xlabel('Input 1 (X1)')
plt.ylabel('Input 2 (X2)')
plt.title('Perceptron Decision Boundary')
plt.legend()
plt.grid(True)
plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.show()
```
